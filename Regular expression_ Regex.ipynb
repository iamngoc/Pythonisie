# Regular expression

A regular expression (Regex) is a sequence of characters that specifies a match pattern in text. Usually such patterns are used by string-searching algorithms for "find" and "find and replace" operations on strings, or for input validation (Wikipedia).

Instead of saying:

“Find the word cat”

you can describe much more general patterns:

“Find every word that starts with c, ends with t, and has one or more letters in between.”

## The most important regex symbols


| Regex    | Meaning                   | Example              |
| -------- | ------------------------- | -------------------- |
| `.`      | Any character             | `c.t` → `cat`, `cut` |
| `^`      | Beginning of string/line  | `^Hello`             |
| `$`      | End of string/line        | `world$`             |
| `*`      | 0 or more occurrences     | `a*`                 |
| `+`      | 1 or more occurrences     | `a+`                 |
| `?`      | 0 or 1 occurrence         | `colou?r`            |
| `{n}`    | Exactly `n` times         | `\d{4}`              |
| `{n,}`   | At least `n` times        | `\d{2,}`             |
| `{n,m}`  | Between `n` and `m` times | `\d{2,4}`            |
| `[]`     | Character class           | `[abc]`              |
| `[^...]` | NOT these characters      | `[^0-9]`             |
| `()`     | Capturing/grouping        | `(cat\|dog)`         |
| `\|`     | OR                        | `cat\|dog`           |
| `\`      | Escape / special sequence | `\.`                 |
| `\d`     | Digit                     | `0-9`                |
| `\D`     | Non-digit                 | `a`, `!`, etc.       |
| `\w`     | Word character            | Letters, digits, `_` |
| `\W`     | Non-word character        | Spaces, punctuation  |
| `\s`     | Whitespace                | Space, tab, newline  |
| `\S`     | Non-whitespace            | Letters, digits      |
| `\b`     | Word boundary             | `\bcat\b`            |
| `\B`     | Not a word boundary       | `\Bcat`              |


CHARACTERS

────────────────────────────

.           - Any character

[abc]       a, b, or c

[^abc]      Anything except a, b, c

[a-z]       Lowercase letter

[A-Z]       Uppercase letter

[0-9]       Digit


SHORTCUTS

────────────────────────────

\d          Digit

\D          Non-digit

\w          Word character

\W          Non-word character

\s          Whitespace

\S          Non-whitespace


QUANTIFIERS

────────────────────────────

"*"          0 or more

"+"          1 or more

?           0 or 1

{n}         Exactly n

{n,}        n or more

{n,m}       n through m


POSITION

────────────────────────────

^           Beginning

$           End

\b          Word boundary

\B          Not a word boundary


GROUPING

────────────────────────────

(...)       Capturing group

(?:...)     Non-capturing group

|           OR


ESCAPING

────────────────────────────

\.          Literal .

\+          Literal +

\*          Literal *

\?          Literal ?

\(          Literal (

\)          Literal )

Python has a built-in package called `re`, which can be used to work with Regular Expressions.

Import the re module:

```python
import re
```

```python
str = "Hi. My name is Mina. What is your name?"

# search() returns a match object if the text obtains that object, and returns the position of the first string containing that object within the text string.

x = re.search("name", str)
start = x.start()
end = x.end()
print (f"name start from {start} to {end}")
```

```python
# findall() returns a list containing all matches
find_a = re.findall("a", str)
print(find_a)
print(len(find_a))
```

```python
# split() returns a list where the string has been split at each match
x = re.split("\\s", str)
print(x)
```

```python
# sub() replaces one or many matches with a string
x = re.sub("\\s", "*", str)

print(x)
```

```python
x = re.sub("\\.\\s", "*", str)

print(x)
```
