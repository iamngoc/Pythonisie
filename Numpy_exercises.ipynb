# Matrix convolution
![CNN Filter](https://austingwalters.com/wp-content/uploads/2019/01/cnn-filter.gif)


Look at the visualization above to see how convolutional filter is applied to a matrix. The filter matrix is often called kernel. The kernal slides over the original matrix from top to bottom, left to right. In each iteration, the kernel is multiplied element-wise with the sliding window in the matrix, and append to the result matrix. Implement this process in the code below

Output dimension formula:

W2 = (W1 - F + 2P) / S + 1

H2 = (H1 - F + 2P) / S + 1

W2, H2: output matrix dimension
W1: input feature map width
H1: input feature map height
F: size of square filter (kernel)
P: Padding
S: Stride


# Explanation of the Convolution Function

## Example: Shape of the Input Data

Assume:

```python
data.shape
```

returns

```python
(5, 28, 28)
```

This means:

| Variable | Meaning |
|----------|---------|
| `batch_size` | 5 images |
| `height` | Each image has a height of 28 pixels |
| `width` | Each image has a width of 28 pixels |

Therefore:

```text
batch_size = 5
height = 28
width = 28
```

---

## Example: Kernel Size

Suppose we have the following kernel:

```python
kernel = [
    [1, 0],
    [0, -1]
]
```

Then:

| Variable | Value |
|----------|------:|
| `height_kernel` | 2 |
| `width_kernel` | 2 |

---

## Computing the Output Dimensions

The output dimensions are computed using the following formulas:

```python
height_out = height - height_kernel + 1
width_out = width - width_kernel + 1
```

These formulas ensure that the kernel is only placed where it completely fits inside the image.

---

## Reserving Memory for the Output

Before performing the convolution, memory is allocated for the output matrix:

```python
out = np.zeros((batch_size, height_out, width_out))
```

Initially, every element is set to zero.

---

## The Three Nested Loops

The convolution is performed using three nested loops.

| Code | Purpose |
|------|---------|
| `for idx in range(batch_size)` | Iterate over each image in the batch |
| `for h in range(height_out)` | Move the kernel from top to bottom (rows) |
| `for w in range(width_out)` | Move the kernel from left to right (columns) |

As a result, the kernel visits **every valid position** in the image.

---

# Intuition: Sliding the Kernel

Suppose we have the following image and kernel.

### Image

```text
1 2 3 4
5 6 7 8
9 1 2 3
4 5 6 7
```

### Kernel (2×2)

```text
1  0
0 -1
```

As the kernel slides over the image, a small window is extracted at each position.

| Step                      | Extracted Window (`A`) |
|---------------------------|------------------------|
| **Step 1** (`h=0`, `w=0`) |   1 2 |
|                           |   5 6 |
|----------|---------|
| **Step 2** (`h=0`, `w=1`) |   2 3 |
|                           | 6 7    |
|----------|---------|
| **Step 3** (`h=0`, `w=2`) |   3 4 |
|                           | 7 8    |

Thus, at every position, a small window having the same size as the kernel is extracted from the image.

---

# Element-wise Multiplication

The extracted window is multiplied element by element with the kernel:

```python
A * kernel
```

Example:

| Window `A` | Kernel | Result (`A * kernel`) | Sum |
|------------|--------|-----------------------|-----|
| 1 2        |1 0       | 1  0              | 1 + 0 + 0 - 6 = -5
  5 6       | 0 -1      | 0 -6              |


Finally, the sum of all products is stored at the current kernel position:

```python
out[idx, h, w] = np.sum(A * kernel)
```

The output matrix is filled one value at a time until every valid kernel position has been processed.

---

# Summary of the Convolution Algorithm

1. Read the dimensions of the input images.
2. Read the dimensions of the kernel.
3. Compute the output dimensions.
4. Allocate memory for the output matrix.
5. Process each image individually.
6. Slide the kernel over every valid position.
7. Extract the corresponding image window.
8. Multiply the window and the kernel element by element.
9. Sum all products.
10. Store the result in the corresponding position of the output matrix.

This procedure is the fundamental operation behind **2D convolution**, which is widely used in **image processing** and **Convolutional Neural Networks (CNNs)**.


````python
from scipy.signal import convolve2d # For checking result only

def convolve(data, kernel):
    # Get shape of data: batch_size (number of images), height and width (shape of image)
    batch_size, height, width = data.shape
    # Get size of kernel
    height_kernel, width_kernel = kernel.shape
    # TODO: Fill in two lines below to calculate shape of output matrix
    height_out = width - width_kernel + 1
    width_out = height - height_kernel + 1
    # Placeholder for output matrix with shape: batch_size, height_out, width_out
    out = np.zeros((batch_size, height_out, width_out))
    
    for idx in range(batch_size): # For each image in input
        for h in range(height_out):# Slide kernel from top to bottom. TODO: Fill in the code
            for w in range(width_out):# Slide kernel from left to right. TODO: Fill in the code
                # TODO: Get the sliding window from the matrix
                A = data[idx, h:h+height_kernel, w:w+width_kernel]
                # TODO: Multiply the window with the kernel element-wise and assign to output in correct index
                out[idx, h, w] = np.sum(A * kernel)
        assert np.max(out[idx] - convolve2d(data[idx], np.fliplr(kernel), mode='valid')) < 1e-10
    return out
````

````python
kernel = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
out = convolve(data, kernel)
show_image(out[0])
````

# Pooling
![Pooling](https://images.deepai.org/glossary-terms/max-pooling-4803093.jpg)


In order to decrease the data size, we implement the max pooling as above (mean pooling can also be used). A kernel of arbitrary size slides over the matrix from left to right, top to bottom. In each sliding window, the maximum pixel is added to the output. The pooling function is implemented below

## Max Pooling

The code:

```python
out[idx] = np.nanmax(
    data[idx].reshape(new_shape),
    axis=(1,3)
)
```

for input block:

|1|7|
|-|-|
|2|3|

The maximum value is 7. So from 4 pixels to 1 pixel

"axis=(1,3)" means the inner dimensions are combined.

the reduced image matrix is returned.

## Average Pooling

average = (1+7+2+3) / 4 = 3. So the average value is 3

````python
def pooling(data, kernel_size = (2, 2), method='max'):
    batch_size, height, width = data.shape[:3]
    height_pool, width_pool = kernel_size
    
    # Number of sliding windows to cover the matrix
    height_out = int(np.ceil(height / height_pool))
    width_out = int(np.ceil(width / width_pool))
    
    new_shape = (height_out, height_pool, width_out, width_pool)
    out = np.zeros((batch_size, height_out, width_out))
    for idx in range(batch_size):
        # In case the matrix size is not divisible by the kernel size, we "pad" the extra with nan value
        im_pad = np.full((height_out * height_pool, width_out * width_pool), np.nan)
        im_pad[:height, :width] = data[idx]
        if method == 'max':
            out[idx] = np.nanmax(data[idx].reshape(new_shape), axis=(1, 3))
        else:
            out[idx] = np.nanmean(data[idx].reshape(new_shape), axis=(1, 3))
    return out
````
