## HTTP-Request methods

Source: https://www.medienpalast.net/glossar/begriff/http-request/

HTTP (Hypertext Transfer Protocol Request) defines a set of request methods to indicate the purpose of the request and what is expected if the request is successful.

The requests come from Client, typically a web browser and would be sent to a Server, to request resources. These resources can be websites, images, videos or have another format, that is available over the internet. The communication between Client and Server takes place over the HTTP-Protocol.


## HTTP-Methods

The most common HTTP methods are:

- GET: to retrieve information from the server

- POST: to serve to transmit data to the server

- PUT: to update the data

- DELETE: the delete resources


### Components

The first fine contains:

- the URL of the requested resource
- the HTTP method
- the HTTP version

Following the first line are the HEADER, which provide additional information about the request, such as the type of client or the accepted formats of the returned data.

POST and PUT requests also include a BODY containing the data to be transmitted. This body can be in various formats such as JSON, XML, or form data. The body is particularly important because it contains the information the server needs to process to fulfill the request.

### Why are HTTP Requests important?

They aim to provide the desired information quickly and efficiently. If an HTTP request fails due to an error or an invalid request, this often has an immediate impact on users: an error message is displayed, and the user does not receive the requested page.

### Error Handling

Servers can return various status codes to indicate the state of a request. For example, a successful GET request returns a 200 status code, while a 404 error indicates that the requested resource could not be found. These status messages are important for developers to identify and resolve errors.

### HTTP Status-Code

Source: https://restfulapi.net/http-status-codes/

HTTP defines these standard status codes that can be used to convey the results of a client’s request. The status codes are divided into five categories.

- 1xx: Informational – Communicates transfer protocol-level information.
- 2xx: Success – Indicates that the client’s request was accepted successfully.
- 3xx: Redirection – Indicates that the client must take some additional action in order to complete their request.
- 4xx: Client Error – This category of error status codes points the finger at clients.
- 5xx: Server Error – The server takes responsibility for these error status codes.



## Beautiful Soup

Beautiful Soup is a python library for pulling data out of HTML and XML files. It works with your favorite parser to provide idiomatic ways of navigating, searching, and modifying the parse tree.

```python
from bs4 import BeautifulSoup
```

```python
# Pass in a string

resp = requests.get("http://data.pr4e.org/romeo.txt")
html = resp.text
soup = BeautifulSoup(html, 'html.parser')
print(soup)
```

```python
# pass in an open filehandle

with open("index.html") as file:
    soup = BeautifulSoup(file, 'html.parser')
    print(soup)

# or

# print(soup.prettify())
```

## Objects components

- Tag

```python
# get element by tag
tag = soup.title
print(tag)
```

- Name

```python
# get name of element
name = tag.name
print(name)
```

```python
# change name of tag
tag.name = "new_title_by_me"
print(tag)
```

- Attributes

```python
# get value of attributes
p = soup("p", "title")
p.result
```

```python
# access dictionary directly as .attrs
attr = soup.a
#print(attr)
a_attr = attr.attrs
print(type(a_attr), a_attr)
```

A tag's attribute can be added, removed and modified. Again, this is done by treating the tag as a dictionary

```python
tag = soup.a
print(tag.attrs)
```

```python
tag["class"] = "amazon"
tag["href"] = "https://www.amazon.de/"
print(tag)
```

```python
del tag["class"]
del tag["href"]
print(tag)
```

### Navigating the tree

Tags may contain ``strings`` and other ``tags``. These elements are the tag’s ``children``. Beautiful Soup provides a lot of different attributes for navigating and iterating over a tag’s children.

- Navigating using tag names

Using a tag name as an attribute will give you only the first tag by that name:

```python
# get head element

head = soup.head
print(head)
```

```python
# get title element
title = soup.new_title_by_me
print(title)
```

```python
# get tag b in body
b = soup.body.b
print(b)
```

```python
# get all the <a> tags
list_a_tags = soup.find_all("a")
print(list_a_tags.result)
```

- **``.contents`` and ``.children``**

```python
cont_html = soup.head.contents
print(len(cont_html))
```

```python
for content in cont_html:
    print(content.name, "<--")
```

```python
body_children = soup.head.children
print(body_children)
for child in body_children:
    print(child.name, '<----')
```

```python
for child in soup.head.descendants:
    print(type(child), child)
```

```python
print(len(list(soup.children)))
print(len(list(soup.descendants)))
```

- **``.strings`` and ``.stripped_strings``**

```python
# Use the .strings generator:
for string in soup.body.strings:
    print(repr(string))
```

```python
# remove white space by using the .stripped_strings generator

for string in soup.body.stripped_strings:
    print(repr(string))
```

### Searching the tree

- a string

```python
# Finds all the <b> tags in the document
soup.find_all('b').result
```

- a regular expression

```python
import re
for tag in soup.find_all(re.compile("^b")):
    print(tag.name)
```

- a list

```python
soup.find_all(["a", "b"]).result
```

- a function

```python
def has_class_but_no_id(tag):
    return tag.has_attr('class') and not tag.has_attr('id')
soup.find_all(has_class_but_no_id).result
```
