# Matplotlib
uses for creating static, interactive and animated data visualizations.

Some common charts:

#### Line plot
shows the relationship between two quantities x and y

```python
from matplotlib import pyplot as plt
X =  [1,2,3,4,5]
Y = [6,7,10,12,15]

plt.plot(X,Y,color = 'r',marker = 'o')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

#### Bar plot
shows the change of one quantity (y) according to another quantity (x)

```python
x = [1,2,5,6,7,9]
y = [11,6,14,3,7,9]

plt.bar(x,y,color='g')
plt.show()
```

#### Histogram
likes bar plot
shows the distribution of an object when x increases

```python
y = [1,2,4,5,6,7,11]
plt.hist(y)
plt.show()
```

#### Scatter plot
draws points corresponding to x, y coordinates
shows how each information fields is distributed
compares to the other two information fields

```python
x = [1,2,5,6,7,9]
y = [11,6,14,3,7,9]
#có thể thêm tọa độ thứ 3 thể hiện nút nhỏ , to

c = [1,3,5,10,6,9]
plt.scatter(x,y,c)
plt.show()
```

#### Pie plot
percentage distribution by circle

```python
labels = 'Vinaphone','vie','mobi','vn'
sizes= [10,10,20,60]
explode = (0,0,0,0)

# explode (adjust the distance from the center)
#shadow indicates whether there is a shadow or not, and startangle indicates which direction the object is rotated.

plt.pie(sizes, explode = explode,labels = labels,autopct = '%1.1f%%',shadow = True,startangle = 90)
plt.show()
```

### To build an object in Matplotlib to draw multiple sub-figures on a single figure
It includes figure, axes, axis, title, label.

Axes are sub-graphs, axis is the axis within axes.

Only one figure can be shown at a time.

```python
# multi-plot

# subplot(nrows, ncols, index, **kwargs) means subplot(number of rows, number of columns, position)
# nrows is the number of rows in axes
# ncols is the number of columns in axes
# figures is like a matrix, where each cell is a graph

from matplotlib import pyplot as plt
import numpy as np
x = np.linspace(-10, 10, 50)
y = x**2
plt.subplot(2, 2, 3)
plt.plot(x, y)
plt.title('First Plot')
plt.subplot(2, 2, 2)
plt.plot(y, x)
plt.title('Second Plot')
plt.show()
```

#### to build an object figure

```python
x = np.linspace(-10,10,50)
y = x**2
fig = plt.figure()
#vị tri x,y do cao ,do rong
ax = fig.add_axes([0,1,2,2])
ax.plot(x,y)
ax.set_xlabel('X label')
ax.set_ylabel('Y label')
ax.set_title('Plot using oop')
ax1 = fig.add_axes([1,0,1,1])
ax1.plot(y,x)
ax1.set_xlabel('X1 label')
ax1.set_ylabel('Y1 label')
ax1.set_title('Plot using oo1')
plt.show()
```

#### Numpy matrix

```python
x = np.linspace(-10,10,50)

y = x**2

fig,axes = plt.subplots(nrows = 3,ncols = 3)
axes[1,1].plot(x,y)
axes[2,2].plot(y,x)

plt.tight_layout() # to leave the axes spaced far enough apart
plt.show()
```

```python
x = np.linspace(-10,10,50)
y = x**2

fig,axes = plt.subplots(nrows = 3,ncols = 3)
axes[1,1].plot(x,y, label = ' x = y')
axes[1,1].plot(y,x, label = 'y =x')
axes[1,1].legend() # add legend
axes[2,2].plot(y,x)
plt.tight_layout()
plt.show()
```
