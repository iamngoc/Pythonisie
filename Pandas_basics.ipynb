Create a Dataframe from a list

```python
#tao tu list
data = [1,2,3,4]
df_from_list = pd.DataFrame(data=data,index=['a','b','c','d'])
#print(df_from_list)
df_from_list
```

```python
# Alternative
data1 =[['A',6],['B',5],['C',3]]
df2 = pd.DataFrame(data = data1,columns = ['col1','col2'])
print(df2.dtypes)
df2
```

Create a Dateframe from a dict or ndarray

```python
#tu dict hoac ndarray
data2 = {'name':['A','B'],'age':['25','30'],'city':['Berlin','Munich']}
df3 = pd.DataFrame(data = data2,index = ['1','2'])
print(df3)
```

```python
#list of dict
data4 = [{'A':1,'B':2},{'A':2,'B':3,'C':5}]
df4 = pd.DataFrame(data = data4)
print(df4)
```

```python
#list of series
data5 = {'col1':pd.Series([1,2,3],index = ['a','b','c']),
         'col2':pd.Series([4,5,6,7],index = ['a','b','c','d'])}
df5 = pd.DataFrame(data = data5,dtype = float)
print(df5)
```

```python
# add a column
#them 1 cột
df7['Total']  = df7['HP'] + df7['Attack']

df7.head()
```

```python
# delete a column
# #xóa 1 cột
df7 = df7.drop(columns=['Total'])
df7.head()
```

```python
# change information in a column
#thay đổi thông tin trong 1 cột
to_replace = {
    'Type 1':'Fairy'
}
value = {
    'Type 1' : 'FAIRY'
}
replace_df = df7.replace(to_replace = to_replace,value = value)
```

```python
replace_df['Type 1'].unique()
```

```python
# turn around
replace_df = df7.replace(to_replace = value,value = to_replace)
replace_df['Type 1'].unique()
```

```python
# search data
# #tìm kiếm data
#ten = grass
df7[df7['Type 1'] == 'Grass']
```

```python
# search data, print out where Type 1 is Grass and Type 2 is Poison
df7[(df7['Type 1'] == 'Grass')&(df7['Type 2'] == 'Poison')]
```

```python
# search with a string, print out where in the Name contains the word "Mega"
#tim voi chuoi
df7[df7['Name'].str.contains('Mega')].head(10)
```

```python
# negation, print out where in the Name NOT contains the word "Mega"

df7[~df7['Name'].str.contains('Mega')]
```

```python
# Statistics
#Count the number of occurrences of species in Type 1
df7.groupby(by ='Type 1').count()
```

```python
df7.groupby(by ='Type 1').sum().head()
```

```python
# use different aggregation
agg = {
    'Attack':'sum',
    'Defense':'count'
}
df7.groupby('Name').agg(agg).head()
```
